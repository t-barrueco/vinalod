```python
from SPARQLWrapper import SPARQLWrapper, JSON
```

```python
import pandas as pd
```

### Classes declared

```python
sparql = SPARQLWrapper("https://publications.europa.eu/webapi/rdf/sparql")

sparql.setQuery("""
PREFIX owl:  <http://www.w3.org/2002/07/owl#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

SELECT DISTINCT * 
WHERE 
{ 
  { ?class a owl:Class }
  UNION
  { ?class a rdfs:Class }

  ?class a ?classType
}
""")

sparql.setReturnFormat(JSON)
results = sparql.query().convert()

results
```

```python
result_value=[]
for r in results["results"]["bindings"]:
    result_value.append({"class":r["class"]["value"],"class_type":r["class"]["type"],"classType":r["classType"]["value"],"classType_type":r["classType"]["type"]})
```

```python
df=pd.DataFrame(result_value)
df
```

```python
df.to_csv("classes.csv",index=False,sep="|",encoding="utf-8")
```

### Properties used

```python
sparql = SPARQLWrapper("https://publications.europa.eu/webapi/rdf/sparql")

sparql.setQuery("""
SELECT DISTINCT ?property
WHERE 
{ ?s ?property ?o . }
""")

sparql.setReturnFormat(JSON)
results = sparql.query().convert()

results
```

```python
result_value=[]
for r in results["results"]["bindings"]:
    result_value.append({"property":r["property"]["value"],"property_type":r["property"]["type"]})
```

```python
df=pd.DataFrame(result_value)
df
```

```python
df.to_csv("properties.csv",index=False,sep="|",encoding="utf-8")
```

### Classes use a property

```python
sparql = SPARQLWrapper("https://publications.europa.eu/webapi/rdf/sparql")

sparql.setQuery("""
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX : <http://publications.europa.eu/ontology/cdm#>

SELECT ?class
WHERE 
{ :event_title rdfs:domain ?class }
""")

sparql.setReturnFormat(JSON)
results = sparql.query().convert()

results
```

```python
sparql = SPARQLWrapper("https://publications.europa.eu/webapi/rdf/sparql")

sparql.setQuery("""
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX : <http://publications.europa.eu/ontology/cdm#>

select ?domain ?range where { 
  :treaty_name rdfs:domain ?domain ;
            rdfs:range ?range .
}
""")

sparql.setReturnFormat(JSON)
results = sparql.query().convert()

results
```

### List of classes ordered per use

```python
sparql = SPARQLWrapper("https://publications.europa.eu/webapi/rdf/sparql")

sparql.setQuery("""
SELECT ?class (COUNT(?s) AS ?instanceTotal)
WHERE
{ ?s a ?class. }
GROUP BY ?class
ORDER BY DESC(?instanceTotal)
""")

sparql.setReturnFormat(JSON)
results = sparql.query().convert()

results
```

```python
result_value=[]
for r in results["results"]["bindings"]:
    result_value.append({"class":r["class"]["value"],"instanceTotal":r["instanceTotal"]["value"]})
```

```python
df=pd.DataFrame(result_value)
df
```

```python
df.to_csv("classes_instanceTotal.csv",index=False,sep="|",encoding="utf-8")
```
