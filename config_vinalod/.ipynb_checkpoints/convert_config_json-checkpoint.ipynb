```python
def convert_json(df):
    json_lines=[]
    for index,row in df.iterrows():
        line={}
        line["class"]=row["CLASS"]
        if(isinstance(row["URL"], str)):
            line["endpoint_url"]=row["URL"]
        else:
            line["endpoint_url"]=""
        if(isinstance(row["QUERY"], str)):
            line["query"]=row["QUERY"]
        else:
            line["query"]=""
        line["hierarchy"]=[]
        line["properties"]=[]
        line["classes_text"]=[]
        line["parameters"]=[]
        line["filters"]=[]
        line["tooltip"]=[]
        line["function"]=""
        line["columns"]=[]
        #for c in row["HIERARCHY"].split("-"):
        if(isinstance(row["HIERARCHY"], str)):
            hierarchy=row["HIERARCHY"].split("-")
            for x in range(len(hierarchy)-1): 
                line["hierarchy"].append({"parent":hierarchy[x],"child":hierarchy[x+1]})
        if(isinstance(row["PROPERTIES"], str)):
            if(row["PROPERTIES"]!=""):
                for p in row["PROPERTIES"].split(";"):
                    properties=p.split("-")
                    line["properties"].append({"class":properties[0],"property":properties[1]})
        line["type"]=row["TYPE"]
        line["option"]=row["OPTION"]
        line["option_text"]=row["OPTION_TEXT"]
        if(isinstance(row["CLASSES"], str)):
            for c in row["CLASSES"].split(";"):
                classes_text=c.split("-")
                if c!="":
                    line["classes_text"].append({"class":classes_text[0],"text":classes_text[1]})  
        if(isinstance(row["PARAMETERS"], str)):
            for p in row["PARAMETERS"].split(";"):
                line["parameters"].append({"property":p})
        if(isinstance(row["FILTERS"], str)):
            for f in row["FILTERS"].split(";"):
                filters=f.split("-")
                if f!="":
                    line["filters"].append({"property":filters[0],"filter_type":filters[1]})
        if(isinstance(row["TOOLTIP"], str)):
            for t in row["TOOLTIP"].split(";"):
                tooltip=t.split("-")
                if t!="":
                    line["tooltip"].append({"property":tooltip[0],"tooltip_text":tooltip[1]})
        if(isinstance(row["FUNCTION"], str)):
            line["function"]=row["FUNCTION"]
        if(isinstance(row["COLUMNS"], str)):
            columns=row["COLUMNS"].split(";")
            for x in range(len(columns)): 
                line["columns"].append(columns[x])
        json_lines.append(line)
    return json_lines
```

```python
import json
json_string=convert_json(df)
json_string = json.dumps(json_string, indent=4)
jsonFile = open("config_basicMode.json", "w")
jsonFile.write(json_string)
jsonFile.close()
```

```python
from SPARQLWrapper import SPARQLWrapper,JSON
json_loaded=json.loads(json_string)
for j in json_loaded:
    #print(j["endpoint_url"])
    sparql = SPARQLWrapper(j["endpoint_url"])
    print(j["query"])
    sparql.setQuery(j["query"])
    sparql.setReturnFormat(JSON)
    results = sparql.query().convert()
    print(results)
```
